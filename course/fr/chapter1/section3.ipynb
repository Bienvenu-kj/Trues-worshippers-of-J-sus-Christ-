```python
from transformers import pipeline
```

### Analyse de sentiments

```python
classifier = pipeline("sentiment-analysis")
classifier("I've been waiting for a HuggingFace course my whole life.")
```

```python
classifier = pipeline("sentiment-analysis", model="tblard/tf-allocine")
classifier("J'ai attendu un cours d'HuggingFace toute ma vie.")
```

Intéressant ! On observe que le résultat est négatif là où pour la version en anglais le résultat est positif.

```python
classifier(
    ["J'ai attendu un cours d'HuggingFace toute ma vie.",
     "Je déteste tellement ça !"]
) # pour classifier plusieurs phrases
```

La phrase "J'ai attendu un cours d'HuggingFace toute ma vie." qui était précedemment négative devient à présent positive.

### Zéro shot classification

```python
classifier = pipeline("zero-shot-classification", model="BaptisteDoyen/camembert-base-xnli")
classifier(
    "C'est un cours sur la bibliothèque Transformers",
    candidate_labels=["éducation", "politique", "affaires"],
)
```

```python
classifier = pipeline("zero-shot-classification", model="Recognai/zeroshot_selectra_medium")
classifier("""Plateforme intelligente  de validation et archivage de projets tutorés, intégrant l'intelligence artificielle. Cas de l'isp bukavu (département d'informatique de gestion)"""
,candidate_labels = ["Intelligence artificielle", "TAL", "Développement web", "Robotique", "Réseaux informatiques", "Développement logiciel"])
```

### Comparaison avec un autre modèle (mDeBERTa-v3)
Testons maintenant avec un modèle multilingue robuste pour voir si les scores de confiance changent.

```python
classifier_compare = pipeline("zero-shot-classification", model="MoritzLaurer/mDeBERTa-v3-base-mnli-xnli")
sequence_to_classify = """Conception d'un système d'alerte contre les inondations : cas de la ville d'UVIRA"""
candidate_labels = ["Intelligence artificielle", "TAL", "Développement web", "Robotique", "Réseaux informatiques", "Développement logiciel"]

results = classifier_compare(sequence_to_classify, candidate_labels=candidate_labels)
display(results)
```

### Optimisation des Étiquettes
Pour réduire l'ambiguïté, nous pouvons utiliser des labels plus explicites qui définissent mieux le périmètre de chaque domaine.

```python
# Exemple de labels optimisés pour éviter la confusion entre infrastructure et développement
labels_optimises = [
    "Algorithmes d'intelligence artificielle",
    "Architecture de réseaux et serveurs",
    "Génie logiciel et programmation",
    "Traitement automatique des langues (TAL)",
    "Conception de sites et applications web",
    "Systèmes robotiques et automatisme"
]

departement_interests_labels = ["Conception de système d'information","Didactique de l'informatique", "Gestion commerciale et administrative", "Informatique de gestion"]

results_opt = classifier_compare(sequence_to_classify, candidate_labels=labels_optimises)
dp_results_opt = classifier_compare(sequence_to_classify, candidate_labels=departement_interests_labels)
display(results_opt)
display(dp_results_opt)
```

### Test de stabilité
Vérifions si le modèle reste stable avec un nouveau texte court (par exemple, sur la création d'une API web) en utilisant les mêmes labels optimisés.

```python
new_sequence = "Développement d'une API REST pour la gestion d'un inventaire avec authentification sécurisée."

results_stability = classifier_compare(new_sequence, candidate_labels=labels_optimises)
display(results_stability)
```

### Conclusion sur l'optimisation
On voit ici l'importance du choix des étiquettes. En enrichissant sémantiquement les labels, on réduit le bruit et on aide le modèle à prendre une décision plus tranchée.

```python
import pandas as pd

# Comparaison rapide des scores pour la première séquence
df_comp = pd.DataFrame({
    'Labels': results_opt['labels'],
    'Score Optimisé': results_opt['scores']
})

display("Répartition des scores avec labels optimisés :", df_comp)
```
